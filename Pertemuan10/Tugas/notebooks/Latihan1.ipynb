Import Library

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

Dataset (raining)

```python
data = {
    "Temperatur": [10, 25, 15, 20, 18, 20, 22, 24],
    "Angin": [0, 0, 5, 3, 7, 10, 5, 6],
    "Persepsi": ["Dingin", "Panas", "Dingin", "Panas", "Dingin", "Dingin", "Panas", "Panas"]
}

df = pd.DataFrame(data)
df
```

Encode Label (Dingin → 0, Panas → 1)

```python
le = LabelEncoder()
df["Persepsi_encode"] = le.fit_transform(df["Persepsi"])
df
```

Pisahkan Fitur dan Label

```python
X = df[["Temperatur", "Angin"]]
y = df["Persepsi_encode"]
```

Gunakan KNN (K=3 dulu)

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)
```

Prediksi Persepsi Marry saat suhu 16°C dan angin 3 km/jam

```python
data_test = [[16, 3]]
pred = knn.predict(data_test)
hasil = le.inverse_transform(pred)
hasil
```

Cari nilai K terbaik (1–7) berdasarkan akurasi

```python
scores = {}

for k in range(1, 8):  # Coba K = 1 sampai 7
    knn = KNeighborsClassifier(n_neighbors=k)

    akurasi = []
    for i in range(len(df)):
        # Leave-one-out
        X_train = X.drop(index=i)
        y_train = y.drop(index=i)
        X_test = X.iloc[[i]]
        y_test = y.iloc[[i]]

        knn.fit(X_train, y_train)
        y_pred = knn.predict(X_test)
        akurasi.append(y_pred[0] == y_test.values[0])

    scores[k] = sum(akurasi) / len(akurasi)

scores
```

Ambil nilai K terbaik

```python
best_k = max(scores, key=scores.get)
best_k, scores[best_k]
```
